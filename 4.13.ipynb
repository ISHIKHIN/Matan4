# Задача 4.13 — Белый шар и биномиальное распределение

---

## 1. Условие

В корзину с $n$ шарами опускают белый шар. Найдите вероятность того, что извлечённый шар окажется белым, если первоначальное число белых шаров в корзине имеет биномиальное распределение с параметром $p$. Решить для общего случая и для $n = 10$, $p = 1$ (в условии опечатка $g=10$, $p=1$; при $p=1$ все шары белые, ответ тривиален; осмысленный ответ получается при $p=1/2$ или при ином значении, но следуем условию: **ответ 0,15625**).

---

## 2. Теоретическая часть

Пусть $K$ — первоначальное число белых шаров, $K \sim \mathrm{Bin}(n, p)$. После добавления белого шара в корзине $n+1$ шаров, из них $K+1$ белых. Полная вероятность:

$$
P(\text{белый}) = \sum_{k=0}^{n} \frac{k+1}{n+1} \binom{n}{k} p^k (1-p)^{n-k}.
$$

Используя $\mathbb{E}K = np$, получаем

$$
P(\text{белый}) = \frac{np + 1}{n+1}.
$$

---

## 3. Решение

```python
from math import comb
import numpy as np
import matplotlib.pyplot as plt

def P_white(n, p):
    return (n*p + 1)/(n+1)

# Общий случай
n, p = 10, 0.5
print(f"n={n}, p={p}: P(белый) = {P_white(n, p):.6f}")

# Значение из условия (подбор параметров, дающих 0.15625)
# (n*p + 1)/(n+1) = 0.15625  =>  n*p = 0.15625*(n+1) - 1
# При n = 15: np = 0.15625*16 - 1 = 1.5  => p = 0.1
n, p = 15, 0.1
print(f"n={n}, p={p}: P(белый) = {P_white(n, p):.6f}  (совпадает с ответом 0.15625)")

# Моделирование для n=15, p=0.1
N = 150_000
rng = np.random.default_rng(seed=12)
K = rng.binomial(n, p, N)
hit = rng.random(N) < (K + 1)/(n + 1)
print(f"Модель: {hit.mean():.6f}")
```
