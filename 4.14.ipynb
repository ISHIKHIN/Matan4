# Задача 4.14 — Две корзины, формула полной вероятности и Байес

---

## 1. Условие

Первая корзина: 6 шаров, число белых $K_1 \sim \mathrm{Bin}(3, 0{,}8)$ (но не более 6).

Вторая корзина: 11 шаров, число белых $K_2 \sim \mathrm{Bin}(4, 0{,}7)$ (но не более 11).

Все шары перекладываются в третью корзину.

1) Найти $P(A)$ — вероятность того, что случайно вынутый из третьей корзины шар белый.

2) Найти $P(H \mid A)$ — вероятность того, что этот шар первоначально был в первой корзине.

**Ответ:** $P(A) = 0{,}3059$; $P(H\mid A) = 0{,}4615$.

---

## 2. Теоретическая часть

### 1) Полная вероятность

Пусть $K_1, K_2$ — числа белых шаров. Тогда в третьей корзине $6 + 11 = 17$ шаров, из них $K_1 + K_2$ белых:

$$
P(A) = \frac{\mathbb{E}[K_1] + \mathbb{E}[K_2]}{17}
= \frac{3\cdot 0{,}8 + 4\cdot 0{,}7}{17}
= \frac{2{,}4 + 2{,}8}{17}
= \frac{5{,}2}{17} \approx 0{,}3059.
$$

### 2) Условная вероятность (Байес)

Шар равновероятно любой из 17. Вероятность, что он белый и из первой корзины:

$$
P(H \cap A) = \frac{\mathbb{E}[K_1]}{17} = \frac{2{,}4}{17}.
$$

Тогда

$$
P(H\mid A) = \frac{P(H\cap A)}{P(A)} = \frac{2{,}4}{5{,}2} = \frac{6}{13} \approx 0{,}4615.
$$

---

## 3. Решение

```python
from math import comb
import numpy as np

# ---------- Аналитика ----------
E1 = 3*0.8
E2 = 4*0.7
P_A = (E1 + E2)/17
P_H_A = E1/17
P_H_given_A = P_H_A / P_A
print(f"1) P(A) = {P_A:.6f}")
print(f"2) P(H|A) = {P_H_given_A:.6f}")

# ---------- Моделирование ----------
N = 150_000
rng = np.random.default_rng(seed=13)
K1 = rng.binomial(3, 0.8, N)
K2 = rng.binomial(4, 0.7, N)
# В третьей корзине 17 шаров; случайно выбираем один
# Вероятность белого:
is_white = rng.random(N) < (K1 + K2)/17
# Для условной вероятности: генерируем выбор шара с меткой корзины
choice = rng.integers(0, 17, N)   # 0..5 — первая, 6..16 — вторая
from_first = choice < 6
# Белый? Если из первой, то белых K1 из 6; из второй — K2 из 11
white = np.where(from_first,
                 choice < K1,
                 (choice - 6) < K2)
print(f"\nМодель: P(A) ≈ {white.mean():.6f}")
print(f"Модель: P(H|A) ≈ {(white & from_first).sum() / white.sum():.6f}")
```
