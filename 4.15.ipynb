# Задача 4.15 — Две точки в прямоугольнике (Монте-Карло)

---

## 1. Условие

В прямоугольной области $|x| \le 20$, $|y| \le 8$ случайным образом выбираются две точки $(x_1, y_1)$ и $(x_2, y_2)$. События:

- $A$ — расстояние между точками меньше 11;
- $B$ — $|x_1 - x_2| < 14$.

Найдите приближённо методом Монте-Карло:

1) $P(A)$;

2) $P(A \mid B)$.

Использовать 100 000 пар точек, ответы округлить до одного знака после запятой.

**Ответ:** 1) $P(A) = 0{,}37$; 2) $P(A\mid B) = 0{,}64$.

---

## 2. Теоретическая часть

Метод Монте-Карло: генерируем $N$ пар точек, равномерно распределённых в прямоугольнике, считаем долю пар, удовлетворяющих условию. Условная вероятность:

$$
P(A\mid B) \approx \frac{N(A \cap B)}{N(B)}.
$$

---

## 3. Решение

```python
import numpy as np

N = 100_000
rng = np.random.default_rng(seed=14)

# Точки в прямоугольнике |x|<=20, |y|<=8
x1 = rng.uniform(-20, 20, N); y1 = rng.uniform(-8, 8, N)
x2 = rng.uniform(-20, 20, N); y2 = rng.uniform(-8, 8, N)

dist = np.hypot(x1 - x2, y1 - y2)
dx = np.abs(x1 - x2)

A = dist < 11
B = dx < 14
A_and_B = A & B

P_A = A.mean()
P_A_given_B = A_and_B.sum() / B.sum()

print(f"1) P(A) ≈ {P_A:.2f}  (ответ: 0.37)")
print(f"2) P(A|B) ≈ {P_A_given_B:.2f}  (ответ: 0.64)")
```
