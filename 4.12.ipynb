# Задача 4.12 — Встреча с различным временем ожидания

---

## 1. Условие

$n$ лиц имеют времена ожидания $w_1, \dots, w_n$ (в долях часа). Моменты прихода равномерны на $[0, 1]$. Найдите вероятности встречи:

а) $n = 2$;

б) $n = 3$;

в) $n = 3$, $w_1 = 1/4$, $w_2 = 1/3$, $w_3 = 1/3$;

г) $n = 4$ (соответствующая формула приведена в условии).

---

## 2. Теоретическая часть

### а) $n = 2$

Встреча: $|x_1 - x_2| \le w_1$ (если ждёт первый) или $|x_1 - x_2| \le w_2$ (если ждёт второй). При разных $w$:

$$
P_2 = 1 - \frac{(1-w_1)^2}{2} - \frac{(1-w_2)^2}{2}.
$$

### б), в), г)

Для $n \ge 3$ используются формулы включений-исключений по «сдвигам» (формулы приведены в условии). Численно удобно считать методом Монте-Карло по определению:

встреча состоялась, если $\exists\, i,j$ такие, что $|x_i - x_j| \le \min(w_i, w_j)$ (каждый ждёт своё время; достаточно, чтобы хотя бы одна пара пересеклась по времени ожидания).

---

## 3. Решение

```python
import numpy as np

def simulate_meeting(w, N=150_000, seed=11):
    """w — список времён ожидания (в долях часа)."""
    rng = np.random.default_rng(seed)
    n = len(w)
    X = rng.uniform(0, 1, (N, n))
    meet = np.zeros(N, dtype=bool)
    for i in range(n):
        for j in range(i+1, n):
            meet |= np.abs(X[:, i] - X[:, j]) <= min(w[i], w[j])
    return meet.mean()

# а) n=2
w1, w2 = 0.3, 0.2
P2_theor = 1 - (1-w1)**2/2 - (1-w2)**2/2
print(f"а) n=2: теор. = {P2_theor:.6f}, модел. = {simulate_meeting([w1, w2]):.6f}")

# в) n=3, w = (1/4, 1/3, 1/3)
w = [1/4, 1/3, 1/3]
# Точное значение по формуле из условия (приведено): 1435/5184
P3_theor = 1435/5184
print(f"в) n=3, w=(1/4,1/3,1/3): теор. = {P3_theor:.6f}, модел. = {simulate_meeting(w):.6f}")

# г) n=4, пример с w = (0.2, 0.25, 0.3, 0.25)
w4 = [0.2, 0.25, 0.3, 0.25]
print(f"г) n=4, w={w4}: модел. = {simulate_meeting(w4):.6f}")
```
