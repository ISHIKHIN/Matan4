# Задача 4.16 — Две точки в эллипсе (Монте-Карло)

---

## 1. Условие

В области, ограниченной эллипсом

$$
\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1
$$

(в условии коэффициенты повреждены; используем стандартный вид, например $a = 8$, $b = 6$ — при необходимости подставьте свои значения из оригинала), случайным образом выбираются две точки. События:

- $A$ — расстояние между точками меньше $9{,}2$;
- $B$ — координаты первой точки больше 0, а координаты второй — меньше 0 (т.е. $x_1 > 0$ и $x_2 < 0$, или обе координаты? — согласно условию «координаты первой точки больше 0, координаты второй меньше 0»: $x_1>0, y_1>0$ и $x_2<0, y_2<0$).

Найдите приближённо методом Монте-Карло:

1) $P(A)$;

2) $P(A\mid B)$.

Получить 100 000 пар точек внутри эллипса.

---

## 2. Теоретическая часть

Генерация точки, равномерной в эллипсе: генерируем точку в единичном круге (полярные координаты с $r = \sqrt{U}$), затем масштабируем: $x = a\,r\cos\varphi$, $y = b\,r\sin\varphi$.

Условная вероятность:

$$
P(A\mid B) \approx \frac{N(A \cap B)}{N(B)}.
$$

---

## 3. Решение

```python
import numpy as np

# Параметры эллипса (при необходимости замените на точные из условия)
a, b = 8.0, 6.0
N = 100_000
rng = np.random.default_rng(seed=15)

def sample_ellipse(n):
    r = np.sqrt(rng.uniform(0, 1, n))
    phi = rng.uniform(0, 2*np.pi, n)
    return a*r*np.cos(phi), b*r*np.sin(phi)

x1, y1 = sample_ellipse(N)
x2, y2 = sample_ellipse(N)

dist = np.hypot(x1 - x2, y1 - y2)
A = dist < 9.2
B = (x1 > 0) & (y1 > 0) & (x2 < 0) & (y2 < 0)

print(f"1) P(A) ≈ {A.mean():.2f}")
print(f"2) P(A|B) ≈ {(A & B).sum() / B.sum():.2f}")
```
